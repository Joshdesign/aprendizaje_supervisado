# 2. Ingeniería de Características (Feature Engineering)

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Cargar los datos
df = pd.read_csv('data/titanic_clean.csv')
X = df [['Pclass', 'Sex', 'Age', 'SibSp', 'Fare', 'Embarked']]

# Definir columnas
numerical_cols = ['Pclass', 'Age', 'SibSp', 'Fare']
categorical_cols = ['Sex', 'Embarked']

# Crear el transformador de columnas
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ]
)

# Ajustar y transformar una muestra para ver el resultado

X_transformed = preprocessor.fit_transform(X)
print("Datos originales")
print(X.iloc[0])
print("/nDatos transformados por el procesador (primer registro) :\n", X_transformed[0])
```
